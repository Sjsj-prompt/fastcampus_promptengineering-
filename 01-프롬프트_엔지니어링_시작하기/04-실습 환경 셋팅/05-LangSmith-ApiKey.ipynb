## ▶︎ LangSmith란?

LangSmith는 **LLM 애플리케이션 개발과 운영을 위한 종합 플랫폼**입니다. 프롬프트 엔지니어링을 처음 시작하는 분들에게는 특히 유용한 도구로, 복잡한 LLM 애플리케이션을 체계적으로 개발하고 관리할 수 있게 도와줍니다.

> 💡 **중요**: LangSmith는 특정 프레임워크에 종속되지 않습니다. LangChain과 함께 사용해도 되고, 다른 프레임워크나 직접 구현한 코드와도 함께 사용할 수 있습니다!

<br>

### LangSmith의 핵심 기능

![LangSmith 화면](images/LangSmith/02-LangSmith-apikey0.svg)


#### 1. 실행 추적 (Tracing)
LLM 애플리케이션이 실행되는 과정을 **단계별로 시각화**해주는 기능입니다.

> **왜 필요한가요?**
  > - LLM 애플리케이션은 입력을 넣으면 출력이 나오지만, 그 내부 동작 과정(예: 프롬프트가 어떻게 처리되고, 어떤 단계로 결과가 생성되는지 등)을 알기 어렵기 때문에 "블랙박스"처럼 느껴집니다.
  > - LangSmith는 이러한 내부 실행 과정을 투명하게 시각화해줍니다.
  > - LangSmith는 LangChain 사용 여부와 관계없이 동급 최고의 추적 기능을 제공합니다.
  
추적은 다음과 같은 문제를 추적하는 데 도움이 될 수 있습니다.
- 예상치 못한 최종 결과
- 에이전트가 루핑되는 이유
- 체인이 예상보다 느린 이유
- 에이전트가 각 단계에서 사용한 토큰 수

<br>

   ![LangSmith 화면](images/LangSmith/02-LangSmith-apikey6.png)


#### 2. 성능 평가 (Evaluation)
여러 프롬프트나 모델의 성능을 **객관적으로 비교**할 수 있는 기능입니다.

**평가할 수 있는 항목들:**
- **정확도**: 답변이 얼마나 정확한가?
- **일관성**: 같은 질문에 항상 비슷한 답변을 하는가?
- **속도**: 응답 시간이 얼마나 걸리는가?
- **비용**: API 사용 비용이 얼마나 드는가?

#### 3. 프롬프트 관리
프롬프트의 **버전 관리와 협업**을 지원하는 기능입니다.

**주요 기능:**
- 프롬프트 변경 이력 자동 저장
- 팀원들과 프롬프트 공유 및 협업
- 실시간 프롬프트 테스트 (Playground)

## ▶︎ LangSmith API Key 발급

1. [LangSmith 플랫폼](https://www.langchain.com/langsmith)에 접속하여 회원가입 혹은 로그인을 진행합니다.
   
    ![LangSmith 화면](images/LangSmith/02-LangSmith-apikey1.png)

<br>

2. 가입 후 이메일 인증하는 절차를 진행해야 합니다. 

<br>

3. Setting > "Personal" > Create API Key를 클릭하여 API 키를 발급 받습니다.

    ![LangSmith 화면](images/LangSmith/02-LangSmith-apikey2.png)

    > ⚠️ **API 키 생성 시 고려사항:**
    > - Description에는 용도를 명확히 적어두세요 (예: "개인 학습용", "회사 프로젝트용")
    > - 나중에 여러 개의 키를 만들 수 있으니 구분하기 쉽게 작성하세요

<br>

4. Description을 입력하고 Create API Key 버튼을 클릭합니다.

   ![LangSmith 화면](images/LangSmith/02-LangSmith-apikey3.png)

<br>

5. 생성된 API 키를 안전한 곳에 **복사** 및 보관하세요.

   > ⚠️ **중요**: 이 키는 다시 확인할 수 없습니다. 꼭 다른 곳에 메모해두도록 합니다.

   > ⚠️ **중요**: GitHub이나 공개된 장소에 절대 업로드하지 마세요!

   ![LangSmith 화면](images/LangSmith/02-LangSmith-apikey4.png)

<br>

6. `.env`에 LangSmith 키 설정
   - `LANGCHAIN_TRACING_V2`: "true"로 설정하면 추적을 시작합니다.
   - `LANGCHAIN_ENDPOINT`: https://api.smith.langchain.com 변경하지 않습니다.
   - `LANGCHAIN_API_KEY`: 이전 단계에서 발급 받은 키를 입력합니다.
   - `LANGCHAIN_PROJECT`: 프로젝트명을 기입하면 해당 프로젝트 그룹으로 모든 실행(Run)이 추적됩니다.
  
    ![LangSmith 화면](images/LangSmith/02-LangSmith-apikey5.png)

## ▶︎ .env 파일 생성 및 API 키 설정

### `.env` 파일 생성 방법

API 키를 안전하게 관리하기 위해 `.env` 파일을 사용합니다.

---

#### 1단계: .env_sample 파일 확인

프로젝트 루트 디렉토리에 `.env_sample` 파일이 준비되어 있습니다.

```bash
fastcampus_promptengineering-/
├── .env_sample      ← 이 파일을 복사해서 사용
├── .gitignore       ← .env 파일이 Git에 올라가지 않도록 설정됨
└── 01-프롬프트_엔지니어링_시작하기/
```

---

#### 2단계: .env 파일 생성

**방법 1: 파일 탐색기에서 (가장 쉬움)**

1. 프로젝트 루트 폴더에서 `.env_sample` 파일을 찾습니다
2. 파일을 복사하여 같은 폴더에 붙여넣기
3. 복사본의 이름을 `.env`로 변경

**방법 2: 터미널/명령 프롬프트에서**

Mac/Linux:
```bash
# 프로젝트 루트 디렉토리에서 실행
cp .env_sample .env
```

Windows:
```bash
copy .env_sample .env
```

**방법 3: VS Code에서**

1. `.env_sample` 파일을 엽니다
2. 파일 내용을 전체 복사 (Ctrl+A, Ctrl+C)
3. 새 파일 생성 (Ctrl+N)
4. 내용 붙여넣기 (Ctrl+V)
5. 파일을 프로젝트 루트에 `.env` 이름으로 저장

---

#### 3단계: API 키 입력

생성한 `.env` 파일을 편집기로 열고, 위에서 발급받은 LangmSmith API 키를 입력합니다.

**수정 전:**
```bash
LANGCHAIN_API_KEY=your-langsmith-api-key-here
LANGCHAIN_TRACING_V2=true
LANGCHAIN_ENDPOINT=https://api.smith.langchain.com
LANGCHAIN_PROJECT=your-project-name
```

**수정 후:**
```bash
LANGCHAIN_API_KEY=...
LANGCHAIN_TRACING_V2=true
LANGCHAIN_ENDPOINT=https://api.smith.langchain.com
LANGCHAIN_PROJECT=test-project
```

> ⚠️ **중요**
> - API 키는 따옴표 없이 입력합니다
> - `=` 좌우에 공백을 넣지 않습니다
> - 한 줄에 하나의 환경변수만 입력합니다

---

#### 4단계: .env 파일 저장

1. `.env` 파일을 저장합니다 (Ctrl+S)
2. 파일이 프로젝트 루트 디렉토리에 있는지 확인
3. 아래 코드 셀로 이동하여 정상 작동 확인

> 💡 **Tip**: 다른 API 키(GPT, Claude 등)도 동일한 방법으로 `.env` 파일에 추가하면 됩니다.

**LangSmith 추적 확인**

LangSmith의 추적(Tracing) 기능을 사용하려면 환경 변수를 통해 설정을 완료해야 합니다. 

설정 방법은 크게 **두 가지**가 있습니다.

**방법 1: .env 파일을 통한 자동 설정 (권장)**

`.env` 파일에 필요한 정보를 모두 입력해 두었다면, Python 코드에서 다음과 같이 로드하기만 하면 됩니다.

```python
from dotenv import load_dotenv

# .env 파일의 환경 변수 로드
load_dotenv()
```

> ✅ **장점**: 한 번 설정해두면 모든 파일에서 동일하게 사용 가능  
> ✅ **보안**: API Key를 코드에 직접 작성하지 않아 안전함

이 방법으로 다음 환경 변수들이 자동으로 설정됩니다:
- `LANGCHAIN_TRACING_V2` → 추적 기능 활성화 여부
- `LANGCHAIN_ENDPOINT` → LangSmith API 엔드포인트 주소
- `LANGCHAIN_PROJECT` → 추적 결과를 저장할 프로젝트명
- `LANGCHAIN_API_KEY` → 발급받은 API 키

---

**방법 2: 코드 내에서 직접 설정하기**

실습 중 **프로젝트명을 바꾸거나**, **일시적으로 추적을 켜고 끄고 싶을 때**는 코드에서 직접 환경 변수를 설정할 수 있습니다.

```python
import os

# LangSmith 추적 활성화
os.environ["LANGCHAIN_TRACING_V2"] = "true"

# LangSmith API 엔드포인트 설정
os.environ["LANGCHAIN_ENDPOINT"] = "https://api.smith.langchain.com"

# 프로젝트명 지정 (예: "프롬프트-실습")
os.environ["LANGCHAIN_PROJECT"] = "프롬프트-실습"

# API Key 설정
os.environ["LANGCHAIN_API_KEY"] = "여기에_발급받은_API_KEY_입력"
```

> 💡 **언제 사용하나요?**
> - 노트북 내에서 실험적으로 다른 프로젝트명을 테스트할 때
> - 특정 셀에서만 추적을 활성화하고 싶을 때
> - 동적으로 설정을 변경해야 할 때

### LangSmith 추적 테스트

이제 LangSmith가 제대로 작동하는지 간단한 LLM 호출로 확인해봅시다. 

아래 코드를 실행한 후 [LangSmith 대시보드](https://smith.langchain.com/)에서 추적 결과를 확인할 수 있습니다.


````python
# LangSmith 추적 테스트 (OpenAI 사용 예시)
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage

# ChatGPT 모델 초기화
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)

# 간단한 질문 실행
messages = [
    HumanMessage(
        content="프롬프트 엔지니어링이란 무엇인가요? 한 문장으로 답변해주세요."
    )
]
response = llm.invoke(messages)

print("응답:", response.content)
print("\n✅ LangSmith 대시보드에서 이 실행 내역을 확인할 수 있습니다!")
print("🔗 https://smith.langchain.com/")
````
