## ▶︎ .env 파일 생성 및 API 키 설정

### `.env` 파일 생성 방법

API 키를 안전하게 관리하기 위해 `.env` 파일을 사용합니다.

---

#### 1단계: .env_sample 파일 확인

프로젝트 루트 디렉토리에 `.env_sample` 파일이 준비되어 있습니다.

```bash
fastcampus_promptengineering-/
├── .env_sample      ← 이 파일을 복사해서 사용
├── .gitignore       ← .env 파일이 Git에 올라가지 않도록 설정됨
└── 01-프롬프트_엔지니어링_시작하기/
```

---

#### 2단계: .env 파일 생성

**방법 1: 파일 탐색기에서 (가장 쉬움)**

1. 프로젝트 루트 폴더에서 `.env_sample` 파일을 찾습니다
2. 파일을 복사하여 같은 폴더에 붙여넣기
3. 복사본의 이름을 `.env`로 변경

**방법 2: 터미널/명령 프롬프트에서**

Mac/Linux:
```bash
# 프로젝트 루트 디렉토리에서 실행
cp .env_sample .env
```

Windows:
```bash
copy .env_sample .env
```

**방법 3: VS Code에서**

1. `.env_sample` 파일을 엽니다
2. 파일 내용을 전체 복사 (Ctrl+A, Ctrl+C)
3. 새 파일 생성 (Ctrl+N)
4. 내용 붙여넣기 (Ctrl+V)
5. 파일을 프로젝트 루트에 `.env` 이름으로 저장

---

#### 3단계: API 키 입력

생성한 `.env` 파일을 편집기로 열고, 위에서 발급받은 Anthropic API 키를 입력합니다.

**수정 전:**
```bash
ANTHROPIC_API_KEY=your-anthropic-api-key-here
```

**수정 후:**
```bash
ANTHROPIC_API_KEY=sk-ant-ap...
```

> ⚠️ **중요**
> - API 키는 따옴표 없이 입력합니다
> - `=` 좌우에 공백을 넣지 않습니다
> - 한 줄에 하나의 환경변수만 입력합니다

---

#### 4단계: .env 파일 저장

1. `.env` 파일을 저장합니다 (Ctrl+S)
2. 파일이 프로젝트 루트 디렉토리에 있는지 확인
3. 아래 코드 셀로 이동하여 정상 작동 확인

> 💡 **Tip**: 다른 API 키(GPT, Gemini 등)도 동일한 방법으로 `.env` 파일에 추가하면 됩니다.

````python
# API KEY를 환경변수로 관리하기 위한 설정 파일
from dotenv import load_dotenv

# API KEY 정보로드
load_dotenv()
````

**API Key 확인**

환경변수(API Key)가 제대로 로드되었는지 확인합니다. (보안을 위해 일부만 출력)

````python
import os

print(f"[API KEY]\n{os.environ['ANTHROPIC_API_KEY'][:-99]}" + "*" * 99)
````

````python
# Anthropic API 테스트

from langchain_anthropic import ChatAnthropic

try:
    llm = ChatAnthropic(model="claude-3-haiku-20240307")
    response = llm.invoke(
        [
            {"role": "system", "content": "당신은 친절한 AI Assistance 입니다."},
            {"role": "user", "content": "say this is test"},
        ]
    )
    print(response.content)
    print(">>> Anthropic API 성공")
except Exception:
    print(">>> Anthropic API 실패")
````
