## ▶︎ Gemini API 키 발급하기

1. Google AI Studio 사이트에 접속하여 로그인합니다.

   ![Gemini 화면](images/Gemini/02-Gemini-apikey1.png)

<br>

2. Google AI Studio에서 `Get API Key` 버튼 클릭

   ![Gemini 화면](images/Gemini/02-Gemini-apikey2.png)

<br>

3. 아래 화면에서 `API 키 만들기` 버튼을 클릭 

   ![Gemini 화면](images/Gemini/02-Gemini-apikey3.png)


<br>

4. `새 키 만들기` 팝업창에서 가져온 프로젝트 선택 > `Create project` 클릭

   ![Gemini 화면](images/Gemini/02-Gemini-apikey4.png)

<br>

5. `새 프로젝트 만들기` 팝업창에서 프로젝트 이름 지정 > 프로젝트 만들기

   ![Gemini 화면](images/Gemini/02-Gemini-apikey5.png)

<br>

6. 다시 `새 키 만들기` 팝업창에서 이전 단계에서 만든 프로젝트 지정 > 키 이름 지정 설정 후 `키 만들기` 클릭

   ![Gemini 화면](images/Gemini/02-Gemini-apikey6.png)
   ![Gemini 화면](images/Gemini/02-Gemini-apikey7.png)

<br>

7. 생성한 키는 아래 목록에서 관리할 수 있습니다. 

   ![Gemini 화면](images/Gemini/02-Gemini-apikey8.png)

<br>

8. 생성된 API 키를 안전한 곳에 **복사** 및 보관하세요.

   > ⚠️ **중요**: 이 키는 다시 확인할 수 없습니다. 꼭 다른 곳에 메모해두도록 합니다.

   > ⚠️ **중요**: GitHub이나 공개된 장소에 절대 업로드하지 마세요!

   ![Gemini 화면](images/Gemini/02-Gemini-apikey9.png)

<br>

9. 복사한 키를 `.env` 파일의 `GOOGLE_API_KEY` 항목에 입력합니다.
   
    <pre>GOOGLE_API_KEY=AIzaSyCSttRIEqEf1_X...</pre>
   
   ![Gemini 화면](images/Gemini/02-Gemini-apikey10.png)

## ▶︎ Gemini API 키 발급 확인 및 설정

```python
# API KEY를 환경변수로 관리하기 위한 설정 파일
from dotenv import load_dotenv

# API KEY 정보로드
load_dotenv()
```

**API Key 확인**

환경변수(API Key)가 제대로 로드되었는지 확인합니다. (보안을 위해 일부만 출력)

```python
import os

print(f"[API KEY]\n{os.environ['GOOGLE_API_KEY'][:-30]}" + "*" * 30)
```

```python
# Gemini API 테스트

from langchain_google_genai import ChatGoogleGenerativeAI

try:
    llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")
    response = llm.invoke(
        [
            {"role": "system", "content": "당신은 친절한 AI Assistance 입니다."},
            {"role": "user", "content": "say this is test"},
        ]
    )
    print(response.content)
    print(">>> Gemini API 성공")
except Exception:
    print(">>> Gemini API 실패")
```
