## ▶︎ Claude API 키 발급하기

1. Claude 플랫폼 사이트에 접속하여 로그인합니다.
   
   ![Claude 로그인 화면](images/Claude/02-Claude-apikey1.png)

<br>

2. Dashboard 페이지에서 `Get API Keys`를 클릭합니다.
   
   ![Dashboard API Keys](images/Claude/02-Claude-apikey2.png)

<br>

3. **Create an API Key** 섹션에서 `+ Create Key`를 클릭합니다.
   
   ![API Key 생성](images/Claude/02-Claude-apikey3.png)

<br>

4. 키 이름을 작성한 후 `Create Key`를 클릭합니다.
   
   ![키 이름 입력](images/Claude/02-Claude-apikey4.png)

<br>

5. 생성된 API 키를 안전한 곳에 **복사** 및 보관하세요.
   
   > ⚠️ **중요**: 이 키는 다시 확인할 수 없습니다. 꼭 다른 곳에 메모해두도록 합니다.
   > ⚠️ **중요**: GitHub이나 공개된 장소에 절대 업로드하지 마세요!
   
   ![API 키 복사](images/Claude/02-Claude-apikey5.png)

<br>

6. 복사한 키를 `.env` 파일의 `ANTHROPIC_API_KEY` 항목에 입력합니다.
   
   <pre>ANTHROPIC_API_KEY=sk-ant-api0...</pre>
   
   ![.env 파일 설정](images/Claude/02-Claude-apikey7.png)

<br>
7. 생성된 키는 API Keys 목록에서 관리할 수 있습니다.
   
   ![API Keys 관리](images/Claude/02-Claude-apikey6.png)

```python
# API KEY를 환경변수로 관리하기 위한 설정 파일
from dotenv import load_dotenv

# API KEY 정보로드
load_dotenv()
```

**API Key 확인**

환경변수(API Key)가 제대로 로드되었는지 확인합니다. (보안을 위해 일부만 출력)

```python
import os

print(f"[API KEY]\n{os.environ['ANTHROPIC_API_KEY'][:-99]}" + "*" * 99)
```

```python
# Anthropic API 테스트

from langchain_anthropic import ChatAnthropic

try:
    llm = ChatAnthropic(model="claude-3-haiku-20240307")
    response = llm.invoke(
        [
            {"role": "system", "content": "당신은 친절한 AI Assistance 입니다."},
            {"role": "user", "content": "say this is test"},
        ]
    )
    print(response.content)
    print(">>> Anthropic API 성공")
except Exception:
    print(">>> Anthropic API 실패")
```
