### 1. 운영체제 확인

먼저 사용 중인 운영체제를 확인해보겠습니다.


````python
# 운영체제 확인
import platform
import sys


def check_os():
    """운영체제 확인"""
    os_name = platform.system()
    python_version = sys.version

    print(f">>> 운영체제: {os_name}")
    print(
        f">>> Python 버전: {python_version}" if python_version else "python 설치 필요"
    )

    if os_name == "Darwin":
        print(">>> macOS 환경입니다.")
        return "macos"
    elif os_name == "Windows":
        print(">>> Windows 환경입니다.")
        return "windows"
    elif os_name == "Linux":
        print(">>> Linux 환경입니다.")
        return "linux"
    else:
        print("알 수 없는 운영체제입니다.")
        return "unknown"


# 운영체제 확인
current_os = check_os()
````

### 4. 라이브러리 설치

가상환경이 활성화된 상태에서 필수 라이브러리를 설치합니다.

> 가상환경이 활성화되어 있는지 확인하세요. 터미널/명령 프롬프트 앞에 `(prompt-env)` or `([가상환경 이름 별도 지정(eng)])`가 표시되어야 합니다.


````python
# 가상환경 활성화 확인
import sys
import os


def check_virtual_env():
    """가상환경 활성화 상태 확인"""
    # 가상환경 활성화 확인
    in_venv = hasattr(sys, "real_prefix") or (
        hasattr(sys, "base_prefix") and sys.base_prefix != sys.prefix
    )

    if in_venv:
        print("가상환경이 활성화되어 있습니다.")
        print(f"Python 경로: {sys.executable}")
        print(f"현재 작업 디렉토리: {os.getcwd()}")
        return True
    else:
        print("가상환경이 활성화되지 않았습니다.")
        print("위의 단계를 따라 가상환경을 활성화한 후 다시 실행하세요.")
        return False


# 가상환경 확인
venv_active = check_virtual_env()
````

**▷ 라이브러리 설치 명령어**

가상환경이 활성화된 터미널/명령 프롬프트에서 다음 명령어를 실행하세요:

방법 1: requirements.txt 사용 (추천)
```bash
pip install -r requirement.txt
```

방법 2: 개별 설치
```bash
pip install langchain langchain-core langchain-openai langchain-anthropic langchain-google-genai python-dotenv jupyter pandas langsmith
```

방법 3: 업그레이드된 pip로 설치
```bash
pip install --upgrade pip
pip install -r requirement.txt
```


````python
# 설치된 라이브러리 확인
def check_installed_packages():
    """설치된 패키지 확인"""
    required_packages = [
        "langchain",
        "langchain_core",
        "langchain_openai",
        "langchain_anthropic",
        "langchain_google_genai",
        "dotenv",
        "jupyter",
        "langsmith",
    ]

    installed_packages = []
    missing_packages = []

    for package in required_packages:
        try:
            if package == "dotenv":
                import dotenv
            else:
                __import__(package)
            installed_packages.append(package)
            print(f"[O] {package}")
        except ImportError:
            missing_packages.append(package)
            print(f"[X] {package}")

    print(f"\n설치 현황:")
    print(f"  설치됨: {len(installed_packages)}개")
    print(f"  누락됨: {len(missing_packages)}개")

    if missing_packages:
        print(f"\n누락된 패키지: {', '.join(missing_packages)}")
        print("위의 설치 명령어를 실행해주세요.")
        return False
    else:
        print(f"\n모든 필수 라이브러리가 설치되었습니다.")
        return True


# 패키지 확인
if venv_active:
    packages_ready = check_installed_packages()
else:
    print("가상환경을 먼저 활성화해주세요.")
````
