## ▶︎ OpenAI 결제 정보 등록 및 Credit balance 추가하기 

1. 웹 브라우저에서 [OpenAI 플랫폼](https://platform.openai.com/)에 접속합니다.
    - **"Sign up"** 또는 **"Log in"** 버튼을 클릭합니다.
    - 우측 상단에서
      - 회원가입을 하지 않았다면 “`Sign Up`” 을 눌러 회원가입을 진행
      - 가입이 된 상태라면 “`Log In`” 버튼을 눌러 로그인

    > ⚠️ https://chatgpt.com/ (chatGPT) 의 회원가입 정보와는 다르니, 주의해주세요! 

    ![openai apikey](images/OpenAI/02-OpenAI-apikey1.png)

<br>

2. 로그인 후 우측 상단의 톱니바퀴(Setting)을 눌러 이동합니다. 

    ![openai apikey](images/OpenAI/02-OpenAI-apikey2.png)

<br>

3. 왼쪽의 “Billing” 메뉴에서 `Add payment details`를 클릭해서 카드를 등록합니다.

    ![Billing > Add payment details  ](images/OpenAI/02-OpenAI-apikey3.png)
    ![Billing > Add payment details  ](images/OpenAI/02-OpenAI-apikey4.png)
    ![Billing > Add payment details  ](images/OpenAI/02-OpenAI-apikey5.png)

<br>

4. 계정에 credit balance 충전 
    > ⚠️ 카드를 등록한 직후, 결제가 되지 않습니다.
    
    > 새로고침을 한 뒤, 일정시간 이후에 다시 시도해보세요!

    ![Billing > Add payment details  ](images/OpenAI/02-OpenAI-apikey6.png)
    ![Billing > Add payment details  ](images/OpenAI/02-OpenAI-apikey7.png)

<br>

5. 충전이 완료됐습니다.

## ▶︎ API 키 발급하기 

1. 로그인 한 다음 아래 링크로 이동합니다. 
   
   🔗 https://platform.openai.com/api-keys

<br>

2. 아래 화면에서 오른쪽 `+ Create new secret key` 를 클릭한다. 

    ![apikey](images/OpenAI/02-OpenAI-apikey9.png)

<br>

3. Name을 필수로 입력하고, 별도 프로젝트가 없다면 `Default Project`로 설정해서 Secret key를 생성합니다.

    ![apikey](images/OpenAI/02-OpenAI-apikey10.png)

<br> 

4. 생성된 키는 안전한 곳에 보관하세요! 
   
    > ⚠️ 깃헙이나 공개된 장소에 업로드 하지 않도록 주의하세요. 
    
    > ⚠️ 생성된 키는 해당 창에서만 확인할 수 있기 때문에 꼭 다른 곳에 메모해두도록 합니다. 

<br>

5. 복사한 키를 `.env` 파일의 `OPENAI_API_KEY` 항목에 입력합니다.
   
   <pre>OPENAI_API_KEY=sk-5K9kI4dia...</pre>
   
   ![.env 파일 설정](images/OpenAI/02-OpenAI-apikey14.png)


6. 생성한 키는 아래 목록에서 관리할 수 있습니다. 

    ![apikey](images/OpenAI/02-OpenAI-apikey11.png)

```python
import os

print(f"[API KEY]\n{os.environ['OPENAI_API_KEY'][:-140]}" + "*" * 140)
```

```python
# OpenAI API 테스트

from langchain_openai import ChatOpenAI

try:
    llm = ChatOpenAI(model="gpt-3.5-turbo")
    response = llm.invoke(
        [
            {"role": "system", "content": "당신은 친절한 AI Assistance 입니다."},
            {"role": "user", "content": "say this is test"},
        ]
    )
    print(response.content)
    print(">>> OpenAI API 성공")
except Exception:
    print(">>> OpenAI API 실패")
```
